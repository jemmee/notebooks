```python
import hashlib

class DedupSystem:
    def __init__(self):
        self.chunk_store = {}  # Stores: {hash: actual_content}
        self.file_index = {}   # Stores: {file_name: [list_of_hashes]}
        self.total_processed_bytes = 0

    def add_file(self, filename, content):
        self.total_processed_bytes += len(content)
        
        # 1. Generate a unique fingerprint (SHA-256)
        fingerprint = hashlib.sha256(content.encode()).hexdigest()
        
        # 2. Check if we already have this data
        if fingerprint not in self.chunk_store:
            # If new, save it
            self.chunk_store[fingerprint] = content
        
        # 3. Save the pointer
        self.file_index[filename] = fingerprint
        print(f"File '{filename}' processed.")

    def get_stats(self):
        stored_bytes = sum(len(c) for c in self.chunk_store.values())
        savings = self.total_processed_bytes - stored_bytes
        ratio = self.total_processed_bytes / stored_bytes if stored_bytes > 0 else 1
        
        return {
            "Original Size": self.total_processed_bytes,
            "Stored Size": stored_bytes,
            "Space Saved": savings,
            "Dedup Ratio": f"{ratio:.2f}:1"
        }

# --- Simulation ---
dedup = DedupSystem()

# Simulate a team backing up their resumes (lots of duplicates!)
dedup.add_file("alice_resume.txt", "Experience: Python, Linux, Astrophysics.")
dedup.add_file("bob_resume.txt", "Experience: Python, Linux, Astrophysics.") # Exact same
dedup.add_file("charlie_resume.txt", "Experience: Java, Windows, Marketing.") # Unique

stats = dedup.get_stats()
print("\n--- Statistics ---")
for key, val in stats.items():
    print(f"{key}: {val}")
```
