```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_hand_touch():
    # Age in months (0 to 8)
    months = np.linspace(0, 8, 100)
    
    # 1. Muscle Strength (Sigmoid growth curve)
    # Strength starts low and plateaus as they gain control
    strength = 1 / (1 + np.exp(-(months - 4) * 2))
    
    # 2. Required Torque to lift arms against gravity
    # This is a constant threshold based on the baby's arm weight
    required_torque = 0.5 
    
    # 3. Coordination/Proprioception factor
    # Even with strength, they need the brain-map to find their hands
    coordination = 1 / (1 + np.exp(-(months - 4.5) * 1.5))
    
    # Combined Success Probability
    success_probability = strength * coordination

    # --- Visualization ---
    plt.figure(figsize=(10, 6))
    plt.plot(months, strength, label="Muscle Strength (Shoulder/Chest)", color='blue', alpha=0.5)
    plt.plot(months, coordination, label="Neurological Coordination", color='green', alpha=0.5)
    plt.plot(months, success_probability, label="Ability to Touch Hands Above Head", color='red', linewidth=3)
    
    plt.axhline(0.5, color='black', linestyle='--', alpha=0.3, label="Threshold for Success")
    plt.axvline(4.5, color='gray', linestyle=':', label="Typical Milestone (4-5 Months)")
    
    plt.title("Developmental Simulation: Reaching Midline & Above Head")
    plt.xlabel("Age (Months)")
    plt.ylabel("Probability / Maturity Scale")
    plt.legend()
    plt.grid(True, alpha=0.2)
    plt.show()

simulate_hand_touch()
```
