```python
import pandas as pd
from datetime import datetime, timedelta

def simulate(start_date_str, years):
    # Orbital cycle constants (in days)
    SYNODIC = 29.530588    # New Moon to New Moon
    DRACONIC = 27.212220   # Crossing the Ecliptic (Node)
    ANOMALISTIC = 27.554551 # Distance from Earth (Perigee/Apogee)
    
    start_date = datetime.strptime(start_date_str, "%Y-%m-%d")
    num_new_moons = int(years * 12.37)
    
    eclipse_list = []
    
    for i in range(num_new_moons):
        t = i * SYNODIC
        current_date = start_date + timedelta(days=t)
        
        # 1. Alignment with the Node (Vertical proximity)
        node_alignment = (t % DRACONIC)
        dist_from_node = min(node_alignment, abs(DRACONIC - node_alignment), 
                             abs(node_alignment - DRACONIC/2))
        
        # 2. Distance from Earth (Apparent Size)
        # Must be near Perigee (dist_from_perigee < ~4 days) to be Total
        perigee_alignment = (t % ANOMALISTIC)
        dist_from_perigee = min(perigee_alignment, abs(ANOMALISTIC - perigee_alignment))
        
        # Check if geometry allows a Total Eclipse
        # Threshold: < 1.4 days from node & < 3.5 days from perigee
        if dist_from_node < 1.4 and dist_from_perigee < 3.5:
            eclipse_list.append({
                "Date": current_date.strftime("%Y-%m-%d"),
                "Type": "Total Solar Eclipse",
                "Year": current_date.year
            })
            
    return pd.DataFrame(eclipse_list)

# Simulate 50 years starting from the March 2015 eclipse
df = simulate("2015-03-20", 50)

print(df.to_string(index=False))
```
