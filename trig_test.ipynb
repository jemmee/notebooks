```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Setup the data: from 0 to 2*pi (one full circle)
# We use 1000 points to make the curves smooth
x = np.linspace(0, 2 * np.pi, 1000)

# 2. Calculate the Trigonometric values
y_sin = np.sin(x)
y_cos = np.cos(x)
y_tan = np.tan(x)

# 3. Handle the Tangent "jump" (Asymptotes)
# Tangent goes to infinity at pi/2 and 3pi/2. We mask these for a cleaner plot.
y_tan[np.abs(y_tan) > 10] = np.nan

# 4. Create the Plot
plt.figure(figsize=(10, 6))

plt.plot(x, y_sin, label='Sine', color='blue', linewidth=2)
plt.plot(x, y_cos, label='Cosine', color='red', linewidth=2)
plt.plot(x, y_tan, label='Tangent', color='green', linestyle='--')

# 5. Formatting the Graph
plt.axhline(0, color='black', linewidth=1) # X-axis
plt.axvline(0, color='black', linewidth=1) # Y-axis
plt.ylim(-3, 3) # Limit y-axis so Tangent doesn't ruin the scale
plt.title("Trigonometric Functions: Sine, Cosine, and Tangent")
plt.xlabel("Radians (0 to 2π)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
```
