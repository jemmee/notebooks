```python
import numpy as np
import faiss

def run_faiss_demo():
    print("--- Starting FAISS Python Demo ---")
    
    # 1. Setup dimensions and data
    d = 64                           # Dimension of vectors
    nb = 10000                       # Database size
    nq = 5                           # Number of queries
    
    # Generate random data (the "database")
    np.random.seed(1234)             
    xb = np.random.random((nb, d)).astype('float32')
    # Generate random queries
    xq = np.random.random((nq, d)).astype('float32')

    # 2. Build the Index (FlatL2 = Exact Euclidean Distance)
    index = faiss.IndexFlatL2(d)   
    print(f"Is index trained? {index.is_trained}")
    
    # 3. Add vectors to the index
    index.add(xb)                  
    print(f"Total vectors in index: {index.ntotal}")

    # 4. Search
    k = 4  # Find the 4 closest neighbors
    distances, indices = index.search(xq, k) 
    
    print("\nSearch Results for first query:")
    print(f"Indices of nearest neighbors: {indices[0]}")
    print(f"Distances: {distances[0]}")

if __name__ == "__main__":
    run_faiss_demo()
```
