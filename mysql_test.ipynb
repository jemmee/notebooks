```python
# kubectl create secret generic mysql-pass --from-literal=password='password123'
#
# kubectl apply -f ./mysql.yaml
#
# kubectl scale deployment mysql --replicas=0

import mysql.connector

# 1. Connect (Note: localhost works because Docker Desktop maps NodePorts to your host)
db = mysql.connector.connect(
    host="localhost",
    port=30306,
    user="root",
    password="password123"
)
cursor = db.cursor()

# 2. CREATE: Setup Database and Table
cursor.execute("CREATE DATABASE IF NOT EXISTS demo_db")
cursor.execute("USE demo_db")
cursor.execute("CREATE TABLE IF NOT EXISTS users (id INT AUTO_INCREMENT PRIMARY KEY, name VARCHAR(255))")

# 3. INSERT (Create)
cursor.execute("INSERT INTO users (name) VALUES (%s)", ("Alice",))
db.commit()
print(f"Created user: Alice")

# 4. READ
cursor.execute("SELECT * FROM users")
print("Current Users:", cursor.fetchall())

# 5. UPDATE
cursor.execute("UPDATE users SET name = %s WHERE name = %s", ("Bob", "Alice"))
db.commit()
print("Updated Alice to Bob")

# 6. DELETE
cursor.execute("DELETE FROM users WHERE name = %s", ("Bob",))
db.commit()
print("Deleted Bob")

cursor.close()
db.close()
```
