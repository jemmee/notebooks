```python
import pandas as pd
from datetime import datetime, timedelta

def simulate_lunar_eclipses(start_date_str, years):
    # Orbital cycle constants (in days)
    SYNODIC = 29.530588    # Full Moon to Full Moon
    DRACONIC = 27.212220   # Time between crossing the ecliptic nodes
    
    # We start near the March 2016 eclipse
    start_date = datetime.strptime(start_date_str, "%Y-%m-%d")
    num_full_moons = int(years * 12.37)
    
    lunar_eclipses = []
    
    for i in range(num_full_moons):
        t = i * SYNODIC
        current_date = start_date + timedelta(days=t)
        
        # Node Alignment (The "Eclipse Season")
        # For a lunar eclipse, the Moon must be near the opposite node of the Sun
        node_alignment = (t % DRACONIC)
        dist_from_node = min(node_alignment, abs(DRACONIC - node_alignment), 
                             abs(node_alignment - DRACONIC/2))
        
        # Thresholds for Lunar Eclipses:
        # Total/Partial (Umbral): dist_from_node < ~0.9 days
        # Penumbral: dist_from_node < ~1.5 days
        
        if dist_from_node < 0.9:
            status = "Total/Partial (Blood Moon)"
        elif dist_from_node < 1.5:
            status = "Penumbral (Subtle Shadow)"
        else:
            continue
            
        lunar_eclipses.append({
            "Date": current_date.strftime("%Y-%m-%d"),
            "Type": status,
            "Node_Distance": round(dist_from_node, 3)
        })
            
    return pd.DataFrame(lunar_eclipses)

# Simulate 30 years from 2016
df_lunar = simulate_lunar_eclipses("2016-03-23", 30)
print(df_lunar.to_string(index=False))
```
