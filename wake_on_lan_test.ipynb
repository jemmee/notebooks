```python
import socket
import struct

def wake_on_lan(mac_address):
    # 1. Format the MAC address
    if len(mac_address) == 17:
        sep = mac_address[2]
        mac_address = mac_address.replace(sep, '')
    
    # 2. Create the "Magic Packet" 
    # (6 bytes of 0xFF followed by the MAC address repeated 16 times)
    raw_mac = bytes.fromhex(mac_address)
    magic_packet = b'\xff' * 6 + raw_mac * 16

    # 3. Send the packet via UDP Broadcast to port 9
    with socket.socket(socket.AF_INET, socket.SOCK_DGRAM) as sock:
        sock.setsockopt(socket.SOL_SOCKET, socket.SO_BROADCAST, 1)
        # We broadcast to the entire subnet
        sock.sendto(magic_packet, ('255.255.255.255', 9))
    
    print(f"Magic Packet sent for {mac_address}")

# Replace with your server's actual NIC MAC address
wake_on_lan('B8:CA:3A:5D:6E:7F')
```
