```python
import yfinance as yf
import numpy as np

# 1. Setup - Let's compare a stable giant (XOM) vs a volatile one (OXY)
tickers = ["XOM", "OXY"]
data = yf.download(tickers, period="1y")['Close']
returns = data.pct_change().dropna()

def calculate_sortino(ticker_returns, rf_annual=0.04):
    # Daily risk-free rate
    rf_daily = rf_annual / 252
    
    # Calculate excess returns
    excess_returns = ticker_returns - rf_daily
    
    # ISOLATE DOWNSIDE: Only look at returns that underperform our target (0)
    downside_returns = excess_returns[excess_returns < 0]
    
    # Calculate Downside Deviation (Standard Deviation of only negative returns)
    downside_deviation = downside_returns.std()
    
    # Annualize: (Mean Excess Return / Downside Dev) * sqrt(252)
    sortino = (excess_returns.mean() / downside_deviation) * np.sqrt(252)
    return sortino

print("--- Sortino Ratios (The 'Good Risk' Metric) ---")
for t in tickers:
    print(f"{t}: {calculate_sortino(returns[t]):.2f}")
```
