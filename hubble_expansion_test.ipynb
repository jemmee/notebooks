```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# 1. Setup Initial Galaxy Positions
n_galaxies = 50
pos = np.random.uniform(-10, 10, (n_galaxies, 2))
H0 = 0.05  # Expansion rate

fig, ax = plt.subplots(figsize=(7, 7))
scat = ax.scatter(pos[:, 0], pos[:, 1], s=20, color='white')
ax.set_facecolor('black')

def update(frame):
    global pos
    # Hubble's Law: v = H0 * d
    # New Position = Old Position + (H0 * Old Position)
    pos = pos + (H0 * pos)
    scat.set_offsets(pos)
    ax.set_xlim(-50, 50)
    ax.set_ylim(-50, 50)
    return scat,

ani = FuncAnimation(fig, update, frames=50, interval=100)
plt.title("The Expanding Universe (Hubble Simulation)")
plt.show()
```
