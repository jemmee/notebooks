```python
import os
import xml.etree.ElementTree as ET

file_path = 'file.xml'
print(f"File exists: {os.path.exists(file_path)}")

# 1. Load and parse the XML
if os.path.getsize(file_path) > 0:
    tree = ET.parse(file_path)
    root = tree.getroot()
else:
    print("Error: The XML file is empty!")

# 2. Accessing attributes of the root
print(f"--- Library: {root.get('name')} ---")

# 3. Iterating through children
for book in root.findall('book'):
    # Accessing attributes
    book_id = book.get('id')
    
    # Accessing child elements using .find()
    title = book.find('title').text
    author = book.find('author').text
    year = book.find('year').text
    
    print(f"[{book_id}] {title} by {author} ({year})")

# 4. Searching with XPath-lite
# Find only the titles of all books
titles = [t.text for t in root.findall('.//title')]
print(f"\nAll Titles: {titles}")
```
