```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_skydive():
    # Constants
    g = 9.81           # Gravity (m/s^2)
    m = 80             # Mass of skydiver (kg)
    rho = 1.225        # Air density (kg/m^3)
    Cd = 1.0           # Drag coefficient (human shape)
    A = 0.7            # Cross-sectional area (m^2)
    
    dt = 0.1           # Time step (seconds)
    t_max = 40         # Total simulation time
    
    # Storage
    time = np.arange(0, t_max, dt)
    velocity = [0]
    acceleration = [g]
    
    for i in range(len(time)-1):
        v = velocity[-1]
        
        # 1. Calculate Drag Force: Fd = 0.5 * rho * v^2 * Cd * A
        drag_f = 0.5 * rho * (v**2) * Cd * A
        
        # 2. Net Force: F_net = F_gravity - F_drag
        net_f = (m * g) - drag_f
        
        # 3. Current Acceleration: a = F/m
        a = net_f / m
        acceleration.append(a)
        
        # 4. Update Velocity: v_new = v_old + a * dt
        v_new = v + a * dt
        velocity.append(v_new)

    # --- Visualization ---
    fig, ax1 = plt.subplots(figsize=(10, 6))

    ax1.set_xlabel('Time (s)')
    ax1.set_ylabel('Speed (m/s)', color='blue')
    ax1.plot(time, velocity, color='blue', linewidth=2, label='Speed')
    ax1.tick_params(axis='y', labelcolor='blue')

    ax2 = ax1.twinx()
    ax2.set_ylabel('Acceleration (m/s²)', color='red')
    ax2.plot(time, acceleration, color='red', linestyle='--', label='Acceleration')
    ax2.tick_params(axis='y', labelcolor='red')

    plt.title("The Journey to Terminal Velocity")
    ax1.grid(True, alpha=0.3)
    plt.show()

simulate_skydive()
```
