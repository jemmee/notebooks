```python
import matplotlib.pyplot as plt

def is_leap(year):
    """The official Gregorian Leap Year logic."""
    if (year % 400 == 0):
        return True
    if (year % 100 == 0):
        return False
    if (year % 4 == 0):
        return True
    return False

def simulate_drift(years_to_simulate):
    actual_year_length = 365.24219
    calendar_year_length = 365.0
    
    drift_days = []
    current_drift = 0
    
    for year in range(1, years_to_simulate + 1):
        # Every year we lose ~0.24 days
        current_drift += (actual_year_length - calendar_year_length)
        
        # If we use the Leap Year rule, we "push" the calendar back 1 day
        if is_leap(2024 + year):
            current_drift -= 1
            
        drift_days.append(current_drift)
    
    return drift_days

# --- Execution & Visualization ---
years = 400
drift = simulate_drift(years)

plt.figure(figsize=(10, 5))
plt.plot(range(years), drift, label='Drift with Gregorian Rules', color='green')
plt.axhline(0, color='black', linestyle='--', alpha=0.5)

plt.title(f"Calendar Stability Over {years} Years")
plt.xlabel("Years into the Future")
plt.ylabel("Drift (Days)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
