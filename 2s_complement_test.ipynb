```python
def to_binary(value, bits=8):
    """Helper to show the binary string of a value at a fixed width."""
    # We use a bitmask (1 << bits) - 1 to simulate fixed-width overflow
    mask = (1 << bits) - 1
    return bin(value & mask)[2:].zfill(bits)

def demo_2s_complement(number, bits=8):
    if number < 0:
        print(f"Representing {number} in {bits}-bit 2's complement:")
        
        # 1. Start with the positive version
        pos = abs(number)
        print(f"Step 0: Positive {pos}        -> {to_binary(pos, bits)}")
        
        # 2. Invert bits (1's complement)
        # In Python, ~x is -(x+1). To show bits, we bitmask.
        inverted = ~pos
        print(f"Step 1: Invert (1s Comp) -> {to_binary(inverted, bits)}")
        
        # 3. Add 1 (2's complement)
        twos_comp = (inverted + 1)
        print(f"Step 2: Add 1            -> {to_binary(twos_comp, bits)}")
        print(f"Final binary string:      {to_binary(twos_comp, bits)}")
    else:
        print(f"Positive {number} in binary: {to_binary(number, bits)} (No flip needed)")

# --- Run the Demo ---
print("--- NEGATIVE FIVE DEMO ---")
demo_2s_complement(-5, bits=8)

print("\n--- THE ADDITION MAGIC ---")
# Show that 10 + (-5) works perfectly in binary
val1 = 10
val2 = -5
mask = 0xFF # 8-bit mask

binary_sum = (val1 + val2) & mask
print(f"Calculation: 10 + (-5)")
print(f"Binary: {to_binary(val1)} + {to_binary(val2)} = {to_binary(binary_sum)}")
print(f"Resulting integer: {binary_sum}")
```
