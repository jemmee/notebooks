```python
import time

def lucas_lehmer(p):
    """Tests if 2^p - 1 is prime."""
    if p == 2:
        return True
    
    # The Mersenne number to test
    m_p = (2 ** p) - 1
    
    # Initialize the sequence with s_0 = 4
    s = 4
    
    # Perform the iteration p-2 times
    # We apply the modulo m_p at every step to keep 's' from 
    # growing to millions of digits.
    for _ in range(p - 2):
        s = (s * s - 2) % m_p
        
    return s == 0

def run_demo():
    print("--- Lucas-Lehmer Primality Test Demo ---")
    # Test a few known Mersenne exponents
    # 7, 13, 17, 19, 31 are known primes. 11 is not.
    test_exponents = [7, 11, 13, 17, 19, 31]
    
    for p in test_exponents:
        start = time.time()
        is_prime = lucas_lehmer(p)
        duration = time.time() - start
        
        status = "PRIME" if is_prime else "COMPOSITE"
        m_val = (2 ** p) - 1
        
        print(f"M_{p} (2^{p}-1 = {m_val}): {status} ({duration:.6f}s)")

if __name__ == "__main__":
    run_demo()
```
