```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Configuration
num_simulations = 50000  # Total number of experiments
samples_per_sim = 10      # Number of dice rolled per experiment (summed together)

# 2. Monte Carlo Simulation
# We generate a large table of random uniform numbers (1 to 6)
# Each row represents one experiment (e.g., rolling 10 dice)
simulations = np.random.randint(1, 7, size=(num_simulations, samples_per_sim))

# 3. Sum the results of each experiment
# This 'averaging' or 'summing' is what triggers the Central Limit Theorem
results = np.sum(simulations, axis=1)

# 4. Visualization
plt.figure(figsize=(10, 6))

# Plot the histogram of the raw Monte Carlo results
count, bins, ignored = plt.hist(results, bins=range(samples_per_sim, (samples_per_sim * 6) + 2), 
                                density=True, color='skyblue', edgecolor='black', alpha=0.7, label='Monte Carlo Samples')

# Calculate the theoretical normal distribution for comparison
mu = results.mean()
sigma = results.std()
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 100)
p = (1 / (np.sqrt(2 * np.pi) * sigma)) * np.exp(-0.5 * ((x - mu) / sigma)**2)

# Overlay the theoretical "Bell Curve"
plt.plot(x, p, 'r', linewidth=2, label='Theoretical Normal Curve')

plt.title(f"Monte Carlo Simulation: Normal Distribution Emerging from {num_simulations} Trials")
plt.xlabel("Sum of 10 Dice Rolls")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
```
