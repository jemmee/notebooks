```python
def llr_test(k, n):
    """Tests if k * 2^n - 1 is prime."""
    # N = k * 2^n - 1
    N = k * (2 ** n) - 1
    
    # 1. Determine starting value 's'
    # This is a simplified selection for s0
    if k == 1:
        if n % 2 == 0:
            s = 4 # Standard LLT
        else:
            s = 4
    elif k % 3 != 0:
        # For many cases where k is not a multiple of 3
        s = 5
    else:
        # For other specific k values, s is determined by 
        # Lucas sequences or specific constants.
        s = 4 

    # 2. Iteration (n - 2 times)
    # This is the core 'S_i = S_{i-1}^2 - 2' logic
    for i in range(n - 2):
        s = pow(s, 2, N) - 2
        # Ensure s remains in the modular range
        if s < 0: s += N

    # 3. Final Check
    return s == 0

# --- Running the Demo ---
# Example: k=7, n=2.  7 * 2^2 - 1 = 27 (Composite)
# Example: k=1, n=7.  1 * 2^7 - 1 = 127 (Prime)
# Example: k=3, n=2.  3 * 2^2 - 1 = 11 (Prime)

test_cases = [(1, 7), (3, 2), (7, 4), (1, 13)]

print(f"{'Expression':<15} | {'Value':<10} | {'Result'}")
print("-" * 40)

for k, n in test_cases:
    val = k * (2**n) - 1
    result = "PRIME" if llr_test(k, n) else "COMPOSITE"
    print(f"{k} * 2^{n} - 1 | {val:<10} | {result}")
```
