```python
import random
import time

class RaftNode:
    def __init__(self, node_id):
        self.node_id = node_id
        self.state = "Follower"
        self.term = 0
        self.votes_received = 0

    def start_election(self):
        self.state = "Candidate"
        self.term += 1
        self.votes_received = 1  # Votes for self
        print(f"Node {self.node_id} became Candidate for Term {self.term}")

    def request_vote(self, candidate_term):
        if candidate_term > self.term:
            self.term = candidate_term
            self.state = "Follower"
            return True # Vote granted
        return False

# --- Simulation ---
nodes = [RaftNode(i) for i in range(3)]
print("--- Raft Election Simulation ---")

# Simulate Leader (Node 0) disappearing
print("Leader (Node 0) timed out!")

# Node 1 notices and starts election
candidate = nodes[1]
candidate.start_election()

# Candidate asks others for votes
for i in [0, 2]:
    if nodes[i].request_vote(candidate.term):
        candidate.votes_received += 1
        print(f"Node {i} voted for Node 1")

if candidate.votes_received >= 2: # Majority of 3
    candidate.state = "Leader"
    print(f"*** Node 1 is now the LEADER for Term {candidate.term} ***")
```
