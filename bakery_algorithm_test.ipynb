```python
import threading
import time
import random

class DeliBakery:
    def __init__(self, num_customers):
        self.num_customers = num_customers
        self.entering = [False] * num_customers
        self.tickets = [0] * num_customers

    def lock(self, customer_id):
        # 1. Take a ticket
        self.entering[customer_id] = True
        self.tickets[customer_id] = max(self.tickets) + 1
        self.entering[customer_id] = False
        
        print(f"Customer {customer_id} took ticket #{self.tickets[customer_id]}")

        # 2. Wait for turn
        for other in range(self.num_customers):
            # Wait if the other customer is currently taking a ticket
            while self.entering[other]:
                pass
            
            # Wait if the other customer has a smaller ticket, or same ticket with smaller ID
            while (self.tickets[other] != 0 and 
                   (self.tickets[other] < self.tickets[customer_id] or 
                    (self.tickets[other] == self.tickets[customer_id] and other < customer_id))):
                pass

    def unlock(self, customer_id):
        print(f"Customer {customer_id} served. Leaving the Deli.")
        self.tickets[customer_id] = 0

def serve_customer(bakery, customer_id):
    bakery.lock(customer_id)
    # Critical Section
    print(f"--- [SERVER] Now helping Customer {customer_id} ---")
    time.sleep(random.uniform(0.1, 0.5)) 
    bakery.unlock(customer_id)

# Running the simulation
deli = DeliBakery(3)
threads = [threading.Thread(target=serve_customer, args=(deli, i)) for i in range(3)]

for t in threads: t.start()
for t in threads: t.join()
```
