```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_jwst_path():
    # 1. Setup the L2 Coordinate (Relative to Earth)
    # L2 is roughly 1.5 million km from Earth
    L2_dist = 1.5e6 
    
    # 2. Generate the Halo Orbit path
    # We use parametric equations to simulate the 3D 'wobble'
    t = np.linspace(0, 2 * np.pi, 200)
    
    # Scale of the halo orbit is massive (~800,000 km wide)
    x_orbit = L2_dist + 1.5e5 * np.cos(t)  # Slight variation in distance from Earth
    y_orbit = 4.0e5 * np.sin(t)           # Wide side-to-side swing
    z_orbit = 2.0e5 * np.sin(2 * t)       # Vertical 'figure-8' component
    
    # 3. Plotting in 3D
    fig = plt.figure(figsize=(10, 7))
    ax = fig.add_subplot(111, projection='3d')
    
    # Represent Earth at the origin
    ax.scatter(0, 0, 0, color='blue', s=100, label='Earth')
    
    # Represent the L2 Point
    ax.scatter(L2_dist, 0, 0, color='red', marker='x', label='L2 Point')
    
    # Plot JWST's Path
    ax.plot(x_orbit, y_orbit, z_orbit, color='orange', linewidth=2, label='JWST Halo Orbit')
    
    # Final Position
    ax.scatter(x_orbit[-1], y_orbit[-1], z_orbit[-1], color='gold', s=50)

    # Aesthetics
    ax.set_title("JWST's 3D Halo Orbit Around L2")
    ax.set_xlabel("X (km from Earth)")
    ax.set_ylabel("Y (Side Swing)")
    ax.set_zlabel("Z (Vertical)")
    ax.legend()
    
    # Set view to see the distance clearly
    ax.view_init(elev=20, azim=-35)
    plt.show()

if __name__ == "__main__":
    simulate_jwst_path()
```
