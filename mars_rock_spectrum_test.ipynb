```python
import numpy as np
import matplotlib.pyplot as plt

wavelengths = np.linspace(200, 500, 1000)
# Simulate noise
spectrum = np.random.normal(0.1, 0.02, 1000)

# Add "Peaks" representing discovered minerals
spectrum += 1.5 * np.exp(-(wavelengths - 280)**2 / 2) # Magnesium
spectrum += 2.0 * np.exp(-(wavelengths - 393)**2 / 2) # Calcium

plt.plot(wavelengths, spectrum, color='red')
plt.title("Curiosity ChemCam Spectrometer Simulation")
plt.xlabel("Wavelength (nm)")
plt.ylabel("Intensity")
plt.grid(True)
plt.show()
```
