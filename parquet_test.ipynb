```python
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

# 1. Create our West Coast Powerhouse Data
# Notice the 'Shohei Ohtani' record—handling 100,000 of these requires efficiency!
data = {
    'Player': ['Shohei Ohtani', 'Logan Webb', 'Mookie Betts', 'Buster Posey'],
    'Team': ['Dodgers', 'Giants', 'Dodgers', 'Giants'],
    'Role': ['Two-Way', 'Pitcher', 'Outfielder', 'Catcher'],
    'Home_Runs': [54, 1, 32, 158],
    'ERA': [3.14, 2.90, None, None]  # None for non-pitchers
}

df = pd.DataFrame(data)

# 2. Save as "Full Proof" Parquet
# We use 'snappy' compression—it's fast and standard for sports analytics
df.to_parquet('mlb_frontier.parquet', engine='pyarrow', compression='snappy')
print("Parquet archive created successfully.")

# 3. The "Inquiry" Logic: Column Projection
# Imagine the file has 100 columns. We only load 'Player' and 'Home_Runs' to save RAM.
df_power = pd.read_parquet('mlb_frontier.parquet', columns=['Player', 'Home_Runs'])
print("\n--- Power Hitter Inquiry ---")
print(df_power)
```
