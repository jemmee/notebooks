```python
import edge_tts

TEXT = "Hello! I am speaking to you using Microsoft's neural voices. Pretty smooth, right?"
VOICE = "en-US-AndrewNeural" # You can also try "en-GB-SoniaNeural"
OUTPUT_FILE = "edge_tts_test.mp3"

# Note: We just define the object and call .save() with 'await' directly
communicate = edge_tts.Communicate(TEXT, VOICE)
await communicate.save(OUTPUT_FILE)

print(f"✅ Audio saved to {OUTPUT_FILE}")

# Optional: Play the audio directly in your notebook cell
from IPython.display import Audio
Audio(OUTPUT_FILE)
```
