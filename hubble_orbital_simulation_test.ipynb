```python
from skyfield.api import load, EarthSatellite

# Hubble's current TLE (Typical data format for satellites)
line1 = '1 20580U 90037B   26012.55182054  .00001234  00000-0  12345-3 0  9991'
line2 = '2 20580  28.4690  30.1234 0003456  12.9876  45.6789 15.09234567123456'

ts = load.timescale()
hubble = EarthSatellite(line1, line2, 'Hubble', ts)

# Get position for "right now"
t = ts.now()
geocentric = hubble.at(t)
subpoint = geocentric.subpoint()

print(f"Hubble Location at {t.utc_strftime()}:")
print(f"Latitude: {subpoint.latitude.degrees:.2f}")
print(f"Longitude: {subpoint.longitude.degrees:.2f}")
print(f"Altitude: {subpoint.elevation.km:.2f} km")
```
