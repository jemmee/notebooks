```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Setup Parameters
L = 1.0  # Width of the box (normalized)
x = np.linspace(0, L, 1000)

# 2. Function to calculate Wavefunction (Psi) and Probability (Psi^2)
def get_quantum_state(n, x, L):
    # The wavefunction for a 1D infinite well: sqrt(2/L) * sin(n*pi*x/L)
    psi = np.sqrt(2/L) * np.sin(n * np.pi * x / L)
    prob_density = psi**2
    return psi, prob_density

# 3. Visualization
plt.figure(figsize=(10, 8))

for n in range(1, 5):
    psi, prob = get_quantum_state(n, x, L)
    
    # Calculate energy (relative n^2)
    energy_level = n**2
    
    # Plot Wavefunction (offset by energy level for clarity)
    plt.plot(x, psi + energy_level, label=f'n={n} (E={energy_level} units)')
    # Fill the probability density (where the electron is likely to be)
    plt.fill_between(x, energy_level, prob + energy_level, alpha=0.2)

plt.axhline(0, color='black', linewidth=1)
plt.title("Quantum Electron: Wavefunctions and Energy Levels (1D Box)")
plt.xlabel("Position in Box (x/L)")
plt.ylabel("Energy / Amplitude")
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
