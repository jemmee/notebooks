```python
import pyttsx3

# Initialize the engine
engine = pyttsx3.init()

# 1. Adjust properties (Optional)
engine.setProperty('rate', 175)    # Speed (default is 200)
engine.setProperty('volume', 1.0)  # Volume (0.0 to 1.0)

# 2. Choose a voice (Optional)
voices = engine.getProperty('voices')

for index, voice in enumerate(voices):
    print(f"Index: {index} | Name: {voice.name} | ID: {voice.id}")
    
# voices[0] is usually Male, voices[1] is usually Female
engine.setProperty('voice', voices[19].id)

# 3. Say the thing
text = "Do not say, ‘I am too young.’ You must go to everyone I send you to and say whatever I command you. Do not be afraid of them, for I am with you and will rescue you."
print("Speaking...")
engine.say(text)

# 4. Process the queue
engine.runAndWait()
```
