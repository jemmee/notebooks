```python
import requests
import json

def get_address_balance(address: str) -> float:
    """Query mempool.space API for confirmed balance (in BTC)."""
    url = f"https://mempool.space/api/address/{address}"
    try:
        resp = requests.get(url, timeout=8)
        if resp.status_code != 200:
            return -1.0  # error
        
        data = resp.json()
        # chain_stats = confirmed, mempool_stats = unconfirmed
        sats_confirmed = data["chain_stats"]["funded_txo_sum"] - data["chain_stats"]["spent_txo_sum"]
        btc = sats_confirmed / 100_000_000
        return btc
    except Exception as e:
        print(f"Error: {e}")
        return -1.0

# Demo
addr = "1A1zP1eP5QGefi2DMPTfTL5SLmv7DivfNa"
balance = get_address_balance(addr)
print(f"Balance of {addr}: {balance:.8f} BTC")   # Should be very small / dust or zero now
```
