```python
import hashlib
import base58  # pip install base58   (or implement base58 yourself)

def is_valid_bitcoin_address(address: str) -> bool:
    """Checks if a Bitcoin address has correct Base58Check format + checksum."""
    try:
        # Decode Base58 → bytes (25 bytes for legacy P2PKH/P2SH)
        decoded = base58.b58decode(address)
        if len(decoded) != 25:
            return False
        
        # Version byte + 20-byte hash + 4-byte checksum
        version_hash = decoded[:21]
        checksum = decoded[21:]
        
        # Double SHA256 of version+hash → first 4 bytes should match checksum
        hash1 = hashlib.sha256(version_hash).digest()
        hash2 = hashlib.sha256(hash1).digest()
        
        return checksum == hash2[:4]
    except:
        return False

# Examples
print(is_valid_bitcoin_address("1A1zP1eP5QGefi2DMPTfTL5SLmv7DivfNa"))     # Satoshi's genesis → True
print(is_valid_bitcoin_address("1A1zP1eP5QGefi2DMPTfTL5SLmv7DivfNb"))     # One char wrong → False
```
