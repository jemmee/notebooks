```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_vulcanoids():
    # Number of hypothetical asteroids
    n_asteroids = 100
    
    # Vulcanoid Zone: 0.07 to 0.21 AU
    radii = np.random.uniform(0.07, 0.21, n_asteroids)
    
    # Kepler's Third Law: T^2 proportional to a^3
    # Period in years = sqrt(a^3) where a is in AU
    periods = np.sqrt(radii**3)
    
    # Simulation Time (1 Earth Year)
    t = np.linspace(0, 1, 1000)
    
    plt.figure(figsize=(8, 8), facecolor='black')
    ax = plt.subplot(111, projection='polar')
    ax.set_facecolor('black')
    
    # Plot the Sun
    ax.scatter(0, 0, color='yellow', s=500, edgecolors='orange', label='Sun')
    
    # Simulate and Plot Orbits
    for r, p in zip(radii, periods):
        # Calculate theta based on time and orbital period
        theta = (2 * np.pi * t / p)
        # Plot a single moment in time for all asteroids to show distribution
        ax.scatter(theta[0], r, color='gray', s=2, alpha=0.6)
        # Draw the orbital paths
        if r == radii[0] or r == radii[-1]: # Draw inner/outer boundaries
            full_orbit = np.linspace(0, 2*np.pi, 100)
            ax.plot(full_orbit, [r]*100, color='white', linestyle='--', alpha=0.3)

    # Plot Mercury for Scale (~0.39 AU)
    mercury_orbit = np.linspace(0, 2*np.pi, 100)
    ax.plot(mercury_orbit, [0.39]*100, color='darkorange', linewidth=2, label='Mercury Orbit')
    
    plt.title("Hypothetical Vulcanoid Population Map", color='white', pad=20)
    ax.set_yticklabels([])
    ax.set_xticklabels([])
    ax.grid(False)
    plt.legend(loc='upper right')
    plt.show()

simulate_vulcanoids()
```
