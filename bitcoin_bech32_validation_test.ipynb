```python
from bech32 import bech32_decode, convertbits

def is_valid_bech32_bitcoin(address: str) -> bool:
    """Full Bech32/Bech32m validation + Bitcoin mainnet check."""
    hrp, data = bech32_decode(address)
    if hrp is None:
        return False  # Invalid checksum or format
    
    # For Bitcoin mainnet SegWit: hrp == 'bc'
    # (testnet would be 'tb' or 'tc')
    if hrp != 'bc':
        return False
    
    # Convert 5-bit words to 8-bit (original witness program)
    try:
        witver, witprog = data[0], convertbits(data[1:], 5, 8, False)
        if witprog is None:
            return False
        # Optional: check program length (20 bytes for P2WPKH, 32 for P2WSH/Taproot)
        if len(witprog) not in (20, 32):
            return False  # Invalid witness program size
    except:
        return False
    
    return True

# Test it
addr = "bc1qar0srrr7xfkvy5l643lydnw9re59gtzzwf5mdq"
print(is_valid_bech32_bitcoin(addr))          # Should be True
print(is_valid_bech32_bitcoin(addr + "x"))    # False (invalid checksum)
```
