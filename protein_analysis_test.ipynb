```python
from Bio.SeqUtils.ProtParam import ProteinAnalysis

# Sample Protein: Human Insulin (A-chain)
# Sequence: GIVEQCCTSICSLYQLENYCN
protein_seq = "GIVEQCCTSICSLYQLENYCN"

# Initialize the analysis tool
analysed_seq = ProteinAnalysis(protein_seq)

# 1. Molecular Weight (in Daltons)
weight = analysed_seq.molecular_weight()

# 2. Isoelectric Point (pI)
# The pH at which the protein is electrically neutral
iso_point = analysed_seq.isoelectric_point()

# 3. Amino Acid Percentages
# Let's see how much Cysteine (C) is present (important for disulfide bonds)
count_amino_acids = analysed_seq.get_amino_acids_percent()

print(f"--- Protein Analysis for: {protein_seq} ---")
print(f"Molecular Weight: {weight:.2f} Da")
print(f"Isoelectric Point: {iso_point:.2f} pH")
print(f"Cysteine Content: {count_amino_acids['C']*100:.1f}%")

# 4. Aromaticity
# Calculates the relative frequency of Phe, Trp, Tyr
print(f"Aromaticity: {analysed_seq.aromaticity():.2f}")
```
