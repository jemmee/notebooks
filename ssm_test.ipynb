```python
import numpy as np

# A very simple State Space 'Update' function
def ssm_step(current_state, new_input, A, B, C):
    # 1. Update the 'Memory' (Hidden State)
    # New State = (Old Memory * A) + (New Input * B)
    next_state = np.dot(A, current_state) + np.dot(B, new_input)
    
    # 2. Generate the 'Output' (Prediction)
    output = np.dot(C, next_state)
    
    return next_state, output

# Setup tiny matrices (In real AI, these are learned)
A = np.array([[0.9]]) # How much of the old memory to keep
B = np.array([[0.1]]) # How much of the new input to care about
C = np.array([[1.0]]) # How to translate state to output

state = np.array([0.0])
inputs = [1, 0, 1, 1, 0] # A sequence of 'data'

print("Processing sequence...")
for i in inputs:
    state, out = ssm_step(state, [i], A, B, C)
    print(f"Input: {i} | Hidden State: {state[0]:.4f} | Output: {out[0]:.4f}")
```
