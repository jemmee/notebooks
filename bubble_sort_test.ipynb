```python
def bubble_sort(array):
    for i in range(len(array) - 1):
        for j in range(len(array) - i - 1):
            if array[j] > array[j + 1]:
                temp = array[j]
                array[j] = array[j + 1]
                array[j + 1] = temp        
    return array

if __name__ == "__main__":
    array = [64, 34, 25, 12, 22, 11, 90]

    print("Original array:")
    print(array)

    array = bubble_sort(array)

    print()
    print("Sorted array:")
    print(array)
```
