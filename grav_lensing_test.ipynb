```python
import numpy as np
import matplotlib.pyplot as plt

def lens_simulation():
    # Grid of the sky
    x = np.linspace(-2, 2, 400)
    y = np.linspace(-2, 2, 400)
    X, Y = np.meshgrid(x, y)
    
    # Einstein Radius (strength of the lens)
    RE = 1.0
    
    # Calculate the deflected positions
    # Beta = Theta - (RE^2 / Theta)
    r = np.sqrt(X**2 + Y**2)
    # Avoid division by zero
    r[r == 0] = 1e-9
    
    # Distort the coordinates
    distortion = 1 - (RE**2 / r**2)
    X_src = X * distortion
    Y_src = Y * distortion
    
    # Create a "Background Source" (a small circle)
    source = np.exp(-((X_src - 0.1)**2 + (Y_src - 0.1)**2) / 0.01)
    
    plt.imshow(source, extent=[-2, 2, -2, 2], cmap='inferno')
    plt.title("Gravitational Lensing: Einstein Ring Simulation")
    plt.colorbar(label='Light Intensity')
    plt.show()

lens_simulation()
```
