```python
import gtts

from gtts import gTTS
from IPython.display import Audio # Only if using Jupyter/Anaconda

print(gtts.lang.tts_langs())

text = "Call to me and I will answer you and tell you great and unsearchable things you do not know."
tts = gTTS(text=text, lang='en', slow=False, tld='com')

# Save the file
tts.save("gtts_test.mp3")

# Play it in a Notebook
Audio("gtts_test.mp3")
```
