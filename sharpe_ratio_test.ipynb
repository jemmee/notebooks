```python
import yfinance as yf
import numpy as np
import pandas as pd

# 1. Choose your Energy Tickers
tickers = ["XOM", "CVX", "SHEL"]

# 2. Download 1 year of data
# Adj Close is critical for Oil stocks because it includes Dividends
# df = yf.download(tickers, period="1y")['Adj Close']
df = yf.download(tickers, period="1y")['Close']

# 3. Calculate Daily Returns
returns = df.pct_change().dropna()

# 4. Define Annual Risk-Free Rate (Current Treasury Yield is ~4%)
rf_annual = 0.04
rf_daily = rf_annual / 252

def calculate_sharpe(ticker_returns):
    excess_returns = ticker_returns - rf_daily
    # Daily Sharpe = Mean of Excess Returns / Std Dev of Returns
    daily_sharpe = excess_returns.mean() / ticker_returns.std()
    # Annualize by multiplying by sqrt of trading days
    return daily_sharpe * np.sqrt(252)

print("\n--- Risk-Adjusted Performance (Sharpe Ratio) ---")
for ticker in tickers:
    sharpe = calculate_sharpe(returns[ticker])
    print(f"{ticker}: {sharpe:.2f}")
```
