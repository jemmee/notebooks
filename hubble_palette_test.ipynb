```python
import matplotlib.pyplot as plt
import numpy as np
from astropy.visualization import make_lupton_rgb

# 1. Create "Fake" Nebula Data (Simulating what Hubble's sensors catch)
# We create some overlapping blobs of gas
x, y = np.ogrid[-5:5:500j, -5:5:500j]
dist = np.sqrt(x**2 + y**2)

# Sulfur II (Red) - A wide, faint outer cloud
s_ii = np.exp(-dist**2 / 10) 
# Hydrogen Alpha (Green) - The main dense gas
h_alpha = np.exp(-dist**2 / 5) 
# Oxygen III (Blue) - Hot, ionized core
o_iii = np.exp(-dist**2 / 2)

# 2. Define the Simulation Function
def run_hubble_simulation(red_channel, green_channel, blue_channel):
    print("Processing 'Hubble Palette' color mapping...")
    
    # This combines the 3 filters into one RGB image
    # Q and stretch control the brightness and contrast
    rgb_image = make_lupton_rgb(red_channel, green_channel, blue_channel, Q=10, stretch=0.05)
    
    plt.figure(figsize=(8, 8))
    plt.imshow(rgb_image)
    plt.title("Simulated Hubble Palette (SHO Mapping)")
    plt.axis('off') # Hide the graph axis numbers
    plt.show() # <--- THIS makes the window pop up!

# 3. CALL the function (This is why it didn't show before!)
run_hubble_simulation(s_ii, h_alpha, o_iii)
```
