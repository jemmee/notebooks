```python
import random

# 1. Public Setup (The Stadium)
# P is a large prime, G is a generator
P = 2000000579  
G = 5

# 2. Prover's Secret (The Key)
# This never leaves the Prover's machine
secret_x = 12345678
public_y = pow(G, secret_x, P)  # This is shared with the Verifier

print(f"Public Key (Y): {public_y}")

def run_zkp_demo():
    # --- PROVER SIDE: Commitment ---
    # Pick a random number 'r' (nonce)
    r = random.randrange(P)
    commitment_u = pow(G, r, P)
    print(f"Prover sends commitment (U): {commitment_u}")

    # --- VERIFIER SIDE: Challenge ---
    # Verifier sends a random challenge 'c'
    challenge_c = random.randrange(P)
    print(f"Verifier sends challenge (C): {challenge_c}")

    # --- PROVER SIDE: Response ---
    # Prover calculates Z = r + (c * x)
    # This proves knowledge of 'x' without revealing it
    response_z = (r + challenge_c * secret_x) % (P - 1)
    print(f"Prover sends response (Z): {response_z}")

    # --- VERIFIER SIDE: Verification ---
    # Check if G^z == U * Y^c (mod P)
    left_side = pow(G, response_z, P)
    right_side = (commitment_u * pow(public_y, challenge_c, P)) % P

    if left_side == right_side:
        print("\n[SUCCESS] Proof Verified! Prover knows the secret.")
    else:
        print("\n[FAILED] Proof rejected.")

run_zkp_demo()
```
