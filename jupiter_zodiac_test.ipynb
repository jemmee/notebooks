```python
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta

def get_jupiter_zodiac(start_date, years=20):
    # Zodiac signs and their starting degrees
    zodiac_signs = [
        "Aries", "Taurus", "Gemini", "Cancer", "Leo", "Virgo", 
        "Libra", "Scorpio", "Sagittarius", "Capricorn", "Aquarius", "Pisces"
    ]
    
    # In Jan 2026, Jupiter is in Gemini (approx 60-90 degrees)
    # Reference: Jan 1, 2026 ~ 75 degrees
    initial_deg = 75 
    orbital_period_days = 11.86 * 365.25
    deg_per_day = 360 / orbital_period_days

    days = np.arange(0, years * 365)
    dates = [start_date + timedelta(days=int(d)) for d in days]
    
    # Current Longitude = Initial + (rate * time)
    # We use % 360 to keep it within the circle
    longitudes = (initial_deg + (deg_per_day * days)) % 360
    
    return dates, longitudes, zodiac_signs

# Execution
start = datetime(2026, 1, 1)
dates, longs, signs = get_jupiter_zodiac(start, years=20)

# Visualization
plt.figure(figsize=(12, 6))
plt.plot(dates, longs, color='peru', linewidth=2, label="Jupiter's Longitude")

# Draw boundaries for Zodiac signs
for i in range(0, 360, 30):
    plt.axhline(i, color='gray', linestyle='--', alpha=0.3)
    sign_name = signs[i // 30]
    plt.text(dates[0], i + 5, sign_name, fontsize=9, fontweight='bold')

plt.title("Jupiter's Movement Through the Zodiac (2026 - 2046)")
plt.ylabel("Ecliptic Longitude (Degrees)")
plt.xlabel("Year")
plt.yticks(range(0, 390, 30))
plt.grid(alpha=0.2)
plt.show()
```
