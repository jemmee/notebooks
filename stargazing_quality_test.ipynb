```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

def simulate_2017_stargazing():
    days = np.arange(1, 366)
    # Simplified Lunar Phase (29.5 day cycle)
    # 0 = Full Moon, 1 = New Moon
    sky_darkness = 0.5 * (1 + np.sin(2 * np.pi * days / 29.5 + 4))
    
    # Highlight specific event peaks
    events = {
        12: "Quadrantids",
        97: "Jupiter Opposition",
        166: "Saturn Opposition",
        225: "Total Eclipse",
        347: "Geminids (Peak Stargazing)"
    }
    
    plt.figure(figsize=(12, 6))
    plt.plot(days, sky_darkness, color='midnightblue', alpha=0.5, label="Sky Darkness (New Moon=1)")
    
    for day, name in events.items():
        color = 'gold' if "Geminids" in name or "Eclipse" in name else 'red'
        plt.axvline(day, color=color, linestyle='--', alpha=0.7)
        plt.text(day, 1.05, name, rotation=45, ha='right', fontsize=9)

    plt.title("2017 Stargazing Quality Simulation")
    plt.xlabel("Day of the Year (2017)")
    plt.ylabel("Darkness Quality (1.0 = Best)")
    plt.fill_between(days, sky_darkness, 0, color='blue', alpha=0.1)
    plt.grid(True, alpha=0.2)
    plt.ylim(0, 1.2)
    plt.show()

simulate_2017_stargazing()
```
