```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Setup Parameters
S0 = 100       # Starting stock price
mu = 0.05      # Drift (Expected annual return: 5%)
sigma = 0.2    # Volatility (Standard deviation: 20%)
T = 1.0        # Time horizon (1 year)
dt = 0.01      # Time step (e.g., roughly 1 day)
N = int(T/dt)  # Number of steps
t = np.linspace(0, T, N)

# 2. Simulate the Randomness (Brownian Motion)
# We generate random 'shocks' from a normal distribution
np.random.seed(42) 
W = np.random.standard_normal(size=N) 
W = np.cumsum(W) * np.sqrt(dt) # Cumulative sum creates the 'path'

# 3. Apply the Stochastic Formula (Itô's Calculus)
# Note the -0.5 * sigma**2 term—that's the 'Itô Correction'!
X = (mu - 0.5 * sigma**2) * t + sigma * W
S = S0 * np.exp(X)

# 4. Visualization
plt.figure(figsize=(10, 5))
plt.plot(t, S, label='Stochastic Path (GBM)', color='#007acc')
plt.axhline(S0, color='black', linestyle='--', alpha=0.5)
plt.title('Simulated Stock Price using Stochastic Calculus')
plt.xlabel('Time (Years)')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
