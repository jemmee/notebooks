```python
import duckdb
import pandas as pd

# 1. Dataset: Miracles in the Gospel of Matthew
miracles_list = [
    {"event": "Healing the Leper", "category": "Healing", "chapter": 8},
    {"event": "Calming the Storm", "category": "Nature", "chapter": 8},
    {"event": "Healing Centurion's Servant", "category": "Healing", "chapter": 8},
    {"event": "Curing Two Gadarene Demoniacs", "category": "Exorcism", "chapter": 8},
    {"event": "Feeding the 5000", "category": "Provision", "chapter": 14},
    {"event": "Walking on Water", "category": "Nature", "chapter": 14},
    {"event": "Feeding the 4000", "category": "Provision", "chapter": 15},
    {"event": "Coin in Fish's Mouth", "category": "Provision", "chapter": 17},
    {"event": "Healing Two Blind Men", "category": "Healing", "chapter": 20}
]

# 2. Wrap the list in a DataFrame
# DuckDB loves DataFrames!
miracles = pd.DataFrame(miracles_list)

con = duckdb.connect(database=':memory:')

print("--- Miracle Distribution by Category ---")
query = """
    SELECT 
        category, 
        count(*) as count,
        list(event) as examples
    FROM miracles 
    GROUP BY category 
    ORDER BY count DESC
"""

# Now it will work perfectly
print(con.execute(query).df())
```
