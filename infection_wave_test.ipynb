```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_pandemic(total_pop=1000, contact_rate=0.3, recovery_rate=0.1, days=150):
    # Initial state
    S = [total_pop - 1]
    I = [1]
    R = [0]
    
    dt = 1 # One day at a time
    
    for t in range(days):
        # Calculate new changes
        # New infections = (rate * S * I) / Total Population
        new_infected = (contact_rate * S[-1] * I[-1]) / total_pop
        # New recoveries = recovery_rate * Infected
        new_recovered = recovery_rate * I[-1]
        
        # Update groups
        S.append(S[-1] - new_infected)
        I.append(I[-1] + new_infected - new_recovered)
        R.append(R[-1] + new_recovered)
        
    return S, I, R

# Run simulation
S, I, R = simulate_pandemic()

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(S, label='Susceptible', color='blue', lw=2)
plt.plot(I, label='Infected (The Wave)', color='red', lw=3)
plt.plot(R, label='Recovered/Immune', color='green', lw=2)

plt.axhline(0, color='black', lw=1)
plt.title("Pandemic Simulation: The Rise and Fall of an Infection")
plt.xlabel("Days Since First Case")
plt.ylabel("Number of People")
plt.legend()
plt.grid(alpha=0.2)
plt.savefig('infection_wave_test.png')
print("Simulation saved!")
```
