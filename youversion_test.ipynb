```python
import getpass
import requests

def get_bible_versions(api_key):
    # The base URL for the YouVersion Platform API
    url = "https://api.youversion.com/v1/bibles/206"
    # url = "https://api.youversion.com/v1/bibles/111"
    # url = "https://api.youversion.com/v1/bibles/111/books/JHN/chapters"
    # url = "https://api.youversion.com/v1/bibles/111/books/JHN/chapters/3/verses"
    # url = "https://api.youversion.com/v1/bibles/111/books/JHN/chapters/3/verses/16"
    
    # Headers required by YouVersion
    headers = {
        "X-YVP-App-Key": api_key,
        "accept": "application/json"
    }
    
    # Optional: Filter for English versions
    params = {
        "language_ranges": "en",
        "page_size": 5
    }

    try:
        response = requests.get(url, headers=headers, params=params)
        
        # Check for authentication errors
        if response.status_code == 401:
            return "Error: Invalid API Key. Please check your YouVersion dashboard."
        
        response.raise_for_status()
        data = response.json()
        
        return data
        
    except requests.exceptions.RequestException as e:
        return f"An error occurred: {e}"

def get_verse(api_key, bible, book, chapter, verse):
    url = "https://api.youversion.com/v1/bibles/" + bible + "/passages/" + book + "." + chapter + "." + verse
    
    # Headers required by YouVersion
    headers = {
        "X-YVP-App-Key": api_key,
        "accept": "application/json"
    }
    
    params = {
        "page_size": 5
    }

    try:
        response = requests.get(url, headers=headers, params=params)
        
        # Check for authentication errors
        if response.status_code == 401:
            return "Error: Invalid API Key. Please check your YouVersion dashboard."
        
        response.raise_for_status()
        data = response.json()
        
        return data.get("content", [])
        
    except requests.exceptions.RequestException as e:
        return f"An error occurred: {e}"

    
if __name__ == "__main__":
    print("--- YouVersion API Version Fetcher ---")
    # user_key = input("Please enter your YouVersion App Key: ").strip()
    user_key = getpass.getpass("Please enter your YouVersion App Key: ").strip()
    
    versions = get_bible_versions(user_key)
    
    if isinstance(versions, list):
        print(f"\nSuccessfully found {len(versions)} versions:")
        for v in versions:
            print(f"- [{v['id']}] {v['abbreviation']}: {v['name']}")
    else:
        print(versions)

    verse = get_verse(user_key, "111", "JHN", "3", "16")
    print(verse)

    verse = get_verse(user_key, "1392", "JHN", "3", "16")
    print(verse)
    
    verse = get_verse(user_key, "1628", "JHN", "3", "16")
    print(verse)
    
    verse = get_verse(user_key, "1638", "JHN", "3", "16")
    print(verse)
        
    verse = get_verse(user_key, "83", "JHN", "3", "16")
    print(verse)
        
    verse = get_verse(user_key, "1638", "JHN", "3", "16")
    print(verse)
            
    verse = get_verse(user_key, "179", "JHN", "3", "16")
    print(verse)
```
