```python
import yaml

# 1. THE DATA: Deeply Nested 2026 F1 Grid
f1_pro_data = [
    {
        "driver": "Lewis Hamilton",
        "team": "Ferrari",
        "technical_specs": {
            "power_unit": {
                "manufacturer": "Ferrari",
                "type": "V6 Hybrid 2026-Spec",
                "energy_recovery": "Dual MGU-K System"
            },
            "chassis": {
                "weight_kg": 768,
                "aero_mode": "Active (Z-Mode Default)"
            }
        },
        "career_history": {
            "titles": 7,
            "key_milestones": ["First Ferrari Pole 2026", "100+ Race Wins"]
        }
    },
    {
        "driver": "Max Verstappen",
        "team": "Red Bull",
        "technical_specs": {
            "power_unit": {
                "manufacturer": "Red Bull Ford",
                "type": "Bio-Fuel V6",
                "manual_override": "Enabled"
            },
            "chassis": {
                "weight_kg": 770,
                "aero_mode": "X-Mode Optimized"
            }
        },
        "career_history": {
            "titles": 3,
            "key_milestones": ["Youngest Race Winner", "2023 Record Holder"]
        }
    }
]

file_name = "yaml_test.yaml"

# 2. PRODUCE: Nested Export
print(f"--- Exporting Professional Data to {file_name} ---")
with open(file_name, "w") as f:
    yaml.dump(f1_pro_data, f, sort_keys=False, default_flow_style=False)

# 3. PARSE: Accessing the Deep Levels
print("\n--- Parsing Deeply Nested Data ---")
with open(file_name, "r") as f:
    grid = yaml.safe_load(f)
    for entry in grid:
        # Accessing Level 2 (technical_specs) and Level 3 (power_unit)
        engine = entry['technical_specs']['power_unit']['manufacturer']
        milestone = entry['career_history']['key_milestones'][0]
        
        print(f"Driver: {entry['driver']}")
        print(f"  > Engine Provider: {engine}")
        print(f"  > Career Highlight: {milestone}")
```
