```python
from web3 import Web3
from datetime import datetime

# ────────────────────────────────────────────────
# Free public RPC endpoints (as of Feb 2026 – these rotate/rotate often)
# Pick one; if one fails, try another.
# Reliable free ones from publicnode, allnodes, etc.
RPC_URLS = [
    "https://ethereum-rpc.publicnode.com",          # Fast & privacy-focused
    "https://rpc.ankr.com/eth",                     # Ankr free tier
    "https://eth.llamarpc.com",                     # Another solid free one
    "https://mainnet.ethereumnodes.com",            # Backup
]

# We'll try them in order until one connects
w3 = None
for url in RPC_URLS:
    try:
        temp_w3 = Web3(Web3.HTTPProvider(url))
        if temp_w3.is_connected():
            w3 = temp_w3
            print(f"Connected to Ethereum via: {url}")
            break
    except Exception as e:
        print(f"Failed {url}: {e}")

if w3 is None:
    print("Could not connect to any public RPC. Try later or use Infura/Alchemy with your own key.")
    exit(1)

# ────────────────────────────────────────────────
# Demo functions
# ────────────────────────────────────────────────

def get_latest_block_info():
    """Get number and basic info about the latest block"""
    block_number = w3.eth.block_number
    block = w3.eth.get_block('latest')
    
    print(f"\nLatest Block Information ({datetime.now().strftime('%Y-%m-%d %H:%M UTC')})")
    print("═══════════════════════════════════════════")
    print(f"Block Number      : {block_number:,}")
    print(f"Block Hash        : {block['hash'].hex()}")
    print(f"Timestamp         : {datetime.fromtimestamp(block['timestamp'])}")
    print(f"Transactions count: {len(block['transactions'])}")
    print(f"Gas Used          : {block['gasUsed']:,}")
    print(f"Gas Limit         : {block['gasLimit']:,}")
    print("═══════════════════════════════════════════")

def get_eth_balance(address: str):
    """Get ETH balance of any address (in ETH, not wei)"""
    try:
        # Ensure checksum format (optional but best practice)
        checksum_addr = w3.to_checksum_address(address)
        balance_wei = w3.eth.get_balance(checksum_addr)
        balance_eth = w3.from_wei(balance_wei, 'ether')
        
        print(f"\nBalance Check for {checksum_addr}")
        print("═══════════════════════════════════════════")
        print(f"Balance           : {balance_eth:,.6f} ETH")
        print(f"Balance (wei)     : {balance_wei:,}")
        print("═══════════════════════════════════════════")
    except Exception as e:
        print(f"Error checking balance: {e}")

# ────────────────────────────────────────────────
# Run the demo!
# ────────────────────────────────────────────────

if __name__ == "__main__":
    print("Ethereum Public Node Demo (using web3.py)")
    print(f"Chain ID: {w3.eth.chain_id} (should be 1 for mainnet)")
    print(f"Current block number: {w3.eth.block_number:,}\n")
    
    # Show latest block
    get_latest_block_info()
    
    # Check a famous public address (Vitalik's often-cited one – changes over time)
    vitalik_example = "0xab5801a7d398351b8be11c439e05c5b3259aec9b"
    get_eth_balance(vitalik_example)
    
    # You can add your own address here too!
    # my_address = "0xYourAddressHere"
    # get_eth_balance(my_address)
    
    print("\nDone! Try replacing the address or adding more features like token balances (ERC-20).")
```
