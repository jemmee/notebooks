```python
import random
import matplotlib.pyplot as plt

def play_st_petersburg():
    payoff = 2
    while random.random() < 0.5:  # Keep flipping tails (50% chance)
        payoff *= 2
    return payoff

def run_simulation(num_games=10000):
    total_winnings = 0
    averages = []
    
    for i in range(1, num_games + 1):
        total_winnings += play_st_petersburg()
        averages.append(total_winnings / i)
        
    return averages

# Execution
n = 100000
results = run_simulation(n)

# Visualization
plt.figure(figsize=(10, 5))
plt.plot(results)
plt.title(f"St. Petersburg Paradox: Running Average over {n} Games")
plt.xlabel("Number of Games Played")
plt.ylabel("Average Payoff ($)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

print(f"Final Average after {n} games: ${results[-1]:.2f}")
```
