```python
import numpy as np
import matplotlib.pyplot as plt

def lennard_jones(r, epsilon=1.0, sigma=1.0):
    # The formula: V = 4 * epsilon * [(sigma/r)^12 - (sigma/r)^6]
    return 4 * epsilon * ((sigma/r)**12 - (sigma/r)**6)

# Simulation range (distance between atoms)
r = np.linspace(0.9, 3.0, 100)
energy = lennard_jones(r)

plt.figure(figsize=(8, 5))
plt.plot(r, energy, color='blue', lw=2)
plt.axhline(0, color='black', ls='--')
plt.ylim(-1.5, 2)
plt.title("Lennard-Jones Potential (Intermolecular Forces)")
plt.xlabel("Distance between atoms (r)")
plt.ylabel("Potential Energy (V)")
plt.grid(True, alpha=0.3)

# Save for your server
plt.savefig('lennard-jones_potential_test.png')
```
